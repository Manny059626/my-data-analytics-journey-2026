````python
# This function takes a name as input
def greet_user(name):
    # name is a variable given to the function
    print("Hello", name)

# Call the function with a value
greet_user("Nada")
greet_user("Alex")
````

````python
# This function adds two numbers
def add_numbers(a, b):
    # We calculate the sum
    total = a + b

    # return sends the result back
    return total

# Store the result in a variable
result = add_numbers(3, 5)

# Print the result
print(result)


# 	•	print() shows something
#   •	return gives something back
````

````python
# This function checks if a number is positive
def check_positive(number):
    # if condition checks a rule
    if number > 0:
        print("The number is positive")
    else:
      print("The number is negative or 0")

# Call the function
check_positive(10)
check_positive(-5)
````

````python
# This function checks a student's grade
def check_grade(score):
    if score >= 90:
        print("Grade: A")
    elif score >= 70:
        print("Grade: B")
    else:
        print("Grade: C")

# Call the function
check_grade(95)
check_grade(80)
check_grade(60)
````

# Introduction to `if` Conditionals in Python

This lesson explains **`if` conditionals** in a very simple and beginner-friendly way.

---

## What is an `if` conditional?

An `if` conditional lets a program **make a decision**.

In simple words, it means:

> **If something is true, do this.**

If it is not true, the program can do something else.

---

## How an `if` statement works

Python checks a **condition**.

- If the condition is **True** → the code runs  
- If the condition is **False** → the code is skipped  

---

## Basic structure of an `if` statement

```python
if condition:
    # code that runs if the condition is True

````python
number = 5

if number > 0:
    print("The number is positive")
````

````python
number = -3

if number > 0:
    print("Positive")
else:
    print("Negative or zero")
````

````python
score = 75

if score >= 90:
    print("Grade A")
elif score >= 70:
    print("Grade B")
else:
    print("Grade C")
````

````python
hour = 10

if hour >= 9 and hour <= 17:
    print("Store is open")
else:
    print("Store is closed")
````

````python
# Rock, Paper, Scissors game

# Player chooses one option
player = input("Choose rock, paper, or scissors: ")

# Computer choice (hardcoded for now to keep it simple)
computer = "scissors"

# Show choices
print("Player chose:", player)
print("Computer chose:", computer)

# First, check if it is a tie
if player == computer:
    print("It's a tie!")

else:
    # Player chooses rock
    if player == "rock":
        if computer == "scissors":
            print("Rock beats scissors. You win!")
        else:
            print("Paper beats rock. You lose.")

    # Player chooses paper
    elif player == "paper":
        if computer == "rock":
            print("Paper beats rock. You win!")
        else:
            print("Scissors beats paper. You lose.")

    # Player chooses scissors
    elif player == "scissors":
        if computer == "paper":
            print("Scissors beats paper. You win!")
        else:
            print("Rock beats scissors. You lose.")

    # If the player enters something invalid
    else:
        print("Invalid choice. Please choose rock, paper, or scissors.")
````
