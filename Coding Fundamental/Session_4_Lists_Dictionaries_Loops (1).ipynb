# Introduction to `if` Conditionals in Python

This lesson explains **`if` conditionals** in a very simple and beginner-friendly way.

---

## What is an `if` conditional?

An `if` conditional lets a program **make a decision**.

In simple words, it means:

> **If something is true, do this.**

If it is not true, the program can do something else.

---

## How an `if` statement works

Python checks a **condition**.

- If the condition is **True** → the code runs  
- If the condition is **False** → the code is skipped  

---

## Basic structure of an `if` statement

```python
if condition:
    # code that runs if the condition is True

````python

# I created a variable called 'number' and I assigned it a value which is 5
number = 7

# the if statement will CHECK first before it run if that condition is TRUE or FALSE.
# It will only execute if the condition is TRUE

if number > 0:
    print("The number is positive")
````

### Session 4 - Lists, Dictionaries, Loop:

### Lists:

````python

# A LIST is a container that stores multiple values in order
# Think of it like a shopping list or a list of scores

# -----------------------------
# Creating lists
# -----------------------------

numbers = [10, 20, 30, 40]          # list of numbers
fruits = ["apple", "banana", "orange"]  # list of strings

print(numbers)   # print the whole list
print(fruits)

# -----------------------------
# Accessing list items
# -----------------------------

print(numbers[0])   # first item (index starts at 0)
print(numbers[1])   # second item
print(numbers[-1])  # last item in the list

# -----------------------------
# Changing list values
# -----------------------------

numbers[0] = 15     # change the first value
print(numbers)

# -----------------------------
# Adding items to a list
# -----------------------------

numbers.append(50)  # add a value to the end of the list
print(numbers)

numbers.insert(1, 12)  # insert a value at position 1
print(numbers)

# -----------------------------
# Removing items from a list
# -----------------------------

numbers.remove(30)  # remove the value 30
print(numbers)

numbers.pop()       # remove the last item
print(numbers)

# -----------------------------
# List length
# -----------------------------

print(len(numbers))  # number of items in the list

# -----------------------------
# Checking if a value exists
# -----------------------------

print(20 in numbers)  # True if 20 is in the list
print(99 in numbers)  # False if 99 is not in the list

# -----------------------------
# Sorting and reversing lists
# -----------------------------

numbers.sort()      # sort values from smallest to largest
print(numbers)

numbers.reverse()   # reverse the order
print(numbers)

# -----------------------------
# Looping through a list
# -----------------------------

for n in numbers:
    print(n)        # print each value one by one

# -----------------------------
# Real example: calculating total and average
# -----------------------------

sales = [100, 200, 150, 300]  # daily sales values

total_sales = 0              # start total at 0

for s in sales:
    total_sales += s         # add each sale to total

average_sales = total_sales / len(sales)  # calculate average

print("Total sales:", total_sales)
print("Average sales:", average_sales)
````

### Dictionaries:

````python
# A DICTIONARY can store lists as values
# This is very common in real data work
# Think of it like a table:
# key = column name
# value = list of data in that column

# -----------------------------
# Creating a dictionary with lists
# -----------------------------

student_scores = {
    "math": [80, 85, 90],
    "science": [78, 88, 92],
    "english": [75, 82, 89]
}

print(student_scores)  # print the whole dictionary

# -----------------------------
# Accessing a list using a key
# -----------------------------

print(student_scores["math"])      # get all math scores
print(student_scores["science"])   # get all science scores

# -----------------------------
# Accessing individual values inside the list
# -----------------------------

print(student_scores["math"][0])   # first math score
print(student_scores["english"][-1])  # last english score

# -----------------------------
# Adding values to a list inside the dictionary
# -----------------------------

student_scores["math"].append(95)   # add a new math score
student_scores["science"].append(90)

print(student_scores)

# -----------------------------
# Updating values inside lists
# -----------------------------

student_scores["english"][1] = 85   # change second english score
print(student_scores)

# -----------------------------
# Removing values from a list
# -----------------------------

student_scores["math"].remove(80)   # remove score 80 from math
print(student_scores)

# -----------------------------
# Adding a new key with a list
# -----------------------------

student_scores["history"] = [70, 75, 80]
print(student_scores)

# -----------------------------
# Removing a key and its list
# -----------------------------

student_scores.pop("science")        # remove science scores
print(student_scores)

# -----------------------------
# Checking if a key exists
# -----------------------------

print("math" in student_scores)      # True
print("biology" in student_scores)   # False

# -----------------------------
# Looping through dictionary and lists
# -----------------------------

for subject, scores in student_scores.items():
    print(subject, scores)

# -----------------------------
# Real example: calculate average score per subject
# -----------------------------

for subject, scores in student_scores.items():
    total = 0

    for s in scores:
        total += s

    average = total / len(scores)
    print(subject, "average:", average)
````

### For Loops:

````python
# A LOOP repeats code automatically
# Instead of writing the same code many times, Python does it for you

# =========================================================
# FOR LOOPS
# =========================================================

# A FOR loop is used when you KNOW what you are looping over
# Example: a list of numbers

numbers = [10, 20, 30, 40]

# How to think about a FOR loop:
# START: first item in the list
# MIDDLE: keep going through each item
# END: stop when there are no more items

for n in numbers:
    print(n)   # this runs once for EACH item in the list

# ---------------------------------------------------------
# FOR loop with range()
# ---------------------------------------------------------

# range(5) creates numbers from 0 to 4
# START: 0
# END: stops before 5
# STEP (increment): +1 by default

for i in range(5):
    print(i)

# ---------------------------------------------------------
# FOR loop with start and end
# ---------------------------------------------------------

# range(start, end)
# START: 1
# END: stops before 6
# STEP: +1

for i in range(1, 6):
    print(i)

# ---------------------------------------------------------
# FOR loop with step (increment)
# ---------------------------------------------------------

# range(start, end, step)
# START: 0
# END: stops before 10
# STEP: increase by 2

for i in range(0, 10, 2):
    print(i)

# =========================================================
# REAL EXAMPLE: TOTAL AND AVERAGE
# =========================================================

sales = [100, 200, 150, 300]

total = 0  # START: total begins at 0

# MIDDLE: loop through each sale
for s in sales:
    total += s  # INCREMENT: add current sale to total

# END: loop stops after last value
average = total / len(sales)

print("Total:", total)
print("Average:", average)

# =========================================================
# WHILE LOOPS
# =========================================================

# A WHILE loop is used when you DO NOT KNOW how many times it will run
# It keeps running as long as a condition is True

# How to think about a WHILE loop:
# START: initialize a variable
# CONDITION: check if we should continue
# INCREMENT: change the variable
# END: condition becomes False

count = 0  # START

while count < 5:  # CONDITION
    print(count)
    count += 1    # INCREMENT (VERY IMPORTANT)

# END: loop stops when count is no longer < 5

# ---------------------------------------------------------
# WHAT HAPPENS IF YOU FORGET INCREMENT?
# ---------------------------------------------------------

# If count += 1 is missing, the loop never ends
# This is called an INFINITE LOOP

# =========================================================
# REAL EXAMPLE: WAITING FOR VALID INPUT
# =========================================================

password = ""  # START

while password != "python":  # CONDITION
    password = input("Enter password: ")  # user input changes value

# END: loop stops when password equals "python"
print("Access granted")

# =========================================================
# COMBINING DICTIONARIES AND LOOPS
# =========================================================

student_scores = {
    "math": [80, 85, 90],
    "english": [75, 82, 88]
}

# Loop through dictionary
for subject, scores in student_scores.items():
    total = 0

    # Loop through list inside dictionary
    for s in scores:
        total += s

    average = total / len(scores)
    print(subject, "average:", average)
````
