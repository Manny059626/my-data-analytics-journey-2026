```python
# 1. Handling Missing Values
customers.isnull().sum()

products.isnull().sum()

stores.isnull().sum()

sales.isnull().sum()

inventory.isnull().sum()
```

```python
# Create copies of original DataFrames for cleaning
customers_clean = customers.copy()
products_clean = products.copy()
stores_clean = stores.copy()
sales_clean = sales.copy()
inventory_clean = inventory.copy()
```

```python
# 3. Basic Information Retrieval# 
```

```python
# For numerical columns with missing values, replace with column mean
for df in [customers_clean, products_clean, stores_clean, sales_clean, inventory_clean]:
    numeric_cols = df.select_dtypes(include="number").columns
    df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
```

```python
# For categorical columns with missing values, replace with most frequent value
for df in [customers_clean, products_clean, stores_clean, sales_clean, inventory_clean]:
    text_cols = df.select_dtypes(include=["object", "string"]).columns
    for col in text_cols:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Convert date columns to datetime and forward fill missing values

customers_clean["registration_date"] = pd.to_datetime(
    customers_clean["registration_date"], errors="coerce"
)
customers_clean["registration_date"] = customers_clean["registration_date"].ffill()

stores_clean["opening_date"] = pd.to_datetime(
    stores_clean["opening_date"], errors="coerce"
)
stores_clean["opening_date"] = stores_clean["opening_date"].ffill()

sales_clean["date"] = pd.to_datetime(
    sales_clean["date"], errors="coerce"
)
sales_clean["date"] = sales_clean["date"].ffill()
```

```python
# Check for duplicate rows
print("Customer duplicates:", customers_clean.duplicated().sum())
print("Product duplicates:", products_clean.duplicated().sum())

# Remove duplicate rows
customers_clean = customers_clean.drop_duplicates()
products_clean = products_clean.drop_duplicates()
```

## 1. What are the key advantages of using Pandas for data cleaning compared to other methods?

Pandas makes data cleaning faster, easier, and more organized. It provides built-in functions for handling missing values, removing duplicates, filtering data, and working with large datasets. Compared to cleaning data manually in spreadsheets, less mistakes, faster and easy to repeat. 

## 2. How would your approach to handling missing values differ if the missing data was not random but had a pattern or meaning?

If the missing data had a pattern or meaning, I would first investigate why the values are missing before replacing them. I would analyze whether the missing values are related to a specific group, location, or time period. Instead of automatically filling in the missing values with the mean or mode, I might keep the missing values, create a new category, or use a different method. 

## 3. What types of data quality issues might not be immediately visible through simple DataFrame inspection methods?

Some data quality issues are not easy to identify by simply looking at the DataFrame. These include inconsistent data formats, misspelled values, duplicate records with slight differences, incorrect dates, outliers, and invalid data. These problems often require additional analysis, validation, or visualization to detect.

## 4. How would you document your data cleaning process to ensure reproducibility?

I would document each step of the data cleaning process by adding comments to my code, explaining why each cleaning step was performed, and keeping the original data unchanged. This makes it easier for others to understand the process and reproduce the same results in the future.

## 5. In what scenarios might it be better to remove rows with missing values rather than imputing them?

It is better to remove rows with missing values when only a small number of records are missing important information or when the missing data could affect the accuracy of the analysis. Removing those rows can help improve data quality and produce more reliable results, especially if replacing the values could introduce bias.
